# Analyses générales

## Quelles sont les attributs que préfèrent les personnes pour le sexe opposé ?

```python
df_attr_sexe_oppose=dataset.loc[:,['gender','attr1_1','sinc1_1','intel1_1','fun1_1','amb1_1','shar1_1']]

cols = ['attr1_1','sinc1_1','intel1_1','fun1_1','amb1_1','shar1_1']


rename_dict = {
    'attr1_1': 'Attractivité',
    'sinc1_1': 'Sincérité',
    'intel1_1': 'Intelligence',
    'fun1_1': 'Fun',
    'amb1_1': 'Ambition',
    'shar1_1': 'Interêts communs'
}

# Calcul des moyennes
df_attr_sexe_oppose[cols] /= 10   # on rapporte la note sur 10
df_h = df_attr_sexe_oppose[df_attr_sexe_oppose['gender'] == 0][cols].mean().rename(index=rename_dict)
df_f = df_attr_sexe_oppose[df_attr_sexe_oppose['gender'] == 1][cols].mean().rename(index=rename_dict)

# Subplots 1 ligne, 2 colonnes
fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=("Attributs que préfèrent les femmes<br> chez les hommes", "Attributs que préfèrent les hommes<br> chez les femmes")
)

# Graphique gauche
fig.add_trace(
    go.Bar(x=df_h.index, y=df_h.values, name="Hommes"),
    row=1,
    col=1
)

# Graphique droite
fig.add_trace(
    go.Bar(x=df_f.index, y=df_f.values, name="Femmes"),
    row=1,
    col=2
)

fig.update_layout(
    title="Moyenne des points de préférence selon le sexe opposé",
    margin=dict(t=160),
    showlegend=False,
    width=800,
    height=450
)

fig.show()
```

Les femmes attribuent plus d'importance à l'intelligence pour les hommes, et les hommes attribuent plus d'importance à l'attractivité chez les femmes. L'ambition n'est pas forcément un point important chez les deux sexes.

##Est ce que ça a un réel impact dans le choix de revoir la personne ?


Est ce que la diminution des matchs n'est pas dûe à une mauvaise évaluation de ses propres préférences ?
On regarde pour chaque sexe la note moyenne donnée par le partenaire sur chaque attribut, uniquement quand le partenaire veut le/la revoir.

```python
df_attr_sexe_oppose_real=dataset.loc[:,['gender','attr_o','sinc_o','intel_o','fun_o','amb_o','shar_o','dec_o']]
df_attr_sexe_oppose_real=df_attr_sexe_oppose_real[df_attr_sexe_oppose_real["dec_o"] == 1]
#quand le partenaire veut revoir la personne


cols = ['attr_o','sinc_o','intel_o','fun_o','amb_o','shar_o']  # ce sont les notes attribuées par le partenaire


rename_dict = {
    'attr_o': 'Attractivité',
    'sinc_o': 'Sincérité',
    'intel_o': 'Intelligence',
    'fun_o': 'Fun',
    'amb_o': 'Ambition',
    'shar_o': 'Interêts communs'
}


# Calcul des moyennes

df_eval_des_hommes = df_attr_sexe_oppose_real[df_attr_sexe_oppose_real['gender'] == 1][cols].mean().rename(index=rename_dict)
df_eval_des_femmes = df_attr_sexe_oppose_real[df_attr_sexe_oppose_real['gender'] == 0][cols].mean().rename(index=rename_dict)


# Subplots 1 ligne, 2 colonnes
fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=("Notes moyennes quand les femmes <br> veulent revoir les hommes",
                    "Notes moyennes quand les hommes<br> veulent revoir les femmes")
)

# Graphique gauche
fig.add_trace(
    go.Bar(x=df_eval_des_hommes.index, y=df_eval_des_hommes.values, name="Hommes"),
    row=1,
    col=1
)

# Graphique droite
fig.add_trace(
    go.Bar(x=df_eval_des_femmes.index, y=df_eval_des_femmes.values, name="Femmes"),
    row=1,
    col=2
)

fig.update_layout(
    title="Notes moyennes attribuées par le partenaire quand il veut le/la revoir",
    margin=dict(t=160),
    showlegend=False,
    width=800,
    height=450
)

fig.show()
```

Les échelles ne sont pas les mêmes que pour les préférences, car ce ne sont plus des points distribués mais des notes pour chaque attribut.
On peut tout de même voir que les femmes choisissent effectivement des hommes qu'elles trouvent intelligents, mais que les hommes choisissent aussi selon ce critère.
Finalement, les personnes choisissent de manière générale des personnes à qui elles donnent des notes élevées sur à peu près tous les attributs.

C'est pourquoi on va s'intéresser aussi aux notes attribuées sur les personnes qu'on décide de ne pas revoir.

```python
df_attr_sexe_oppose_dec0=dataset.loc[:,['gender','attr_o','sinc_o','intel_o','fun_o','amb_o','shar_o','dec_o']]
df_attr_sexe_oppose_dec0=df_attr_sexe_oppose_dec0[df_attr_sexe_oppose_dec0["dec_o"] == 0]
# quand le partenaire ne veut pas revoir la personne


cols = ['attr_o','sinc_o','intel_o','fun_o','amb_o','shar_o']  # ce sont les notes attribuées par le partenaire


rename_dict = {
    'attr_o': 'Attractivité',
    'sinc_o': 'Sincérité',
    'intel_o': 'Intelligence',
    'fun_o': 'Fun',
    'amb_o': 'Ambition',
    'shar_o': 'Interêts communs'
}


# Calcul des moyennes

df_eval_des_hommes = df_attr_sexe_oppose_dec0[df_attr_sexe_oppose_dec0['gender'] == 1][cols].mean().rename(index=rename_dict)
df_eval_des_femmes = df_attr_sexe_oppose_dec0[df_attr_sexe_oppose_dec0['gender'] == 0][cols].mean().rename(index=rename_dict)


# Subplots 1 ligne, 2 colonnes
fig = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=("Notes moyennes quand les femmes <br> ne veulent pas revoir les hommes",
                    "Notes moyennes quand les hommes<br> ne veulent pas revoir les femmes")
)

# Graphique gauche
fig.add_trace(
    go.Bar(x=df_eval_des_hommes.index, y=df_eval_des_hommes.values, name="Hommes"),
    row=1,
    col=1
)

# Graphique droite
fig.add_trace(
    go.Bar(x=df_eval_des_femmes.index, y=df_eval_des_femmes.values, name="Femmes"),
    row=1,
    col=2
)

fig.update_layout(
    title="Notes moyennes attribuées par le partenaire quand il ne veut pas le/la revoir",
    margin=dict(t=160),
    showlegend=False,
    width=800,
    height=450
)

fig.show()
```

Conclusion : les personnes de manière générale ont du mal à évaluer leurs préférences, ou cela est plus compliqué et dépend de plein d'autres paramètres.
Les personnes à qui on attribut des bonnes notes partout ont plus de chances d'avoir un match, mais cela ne fonctionne pas vraiment par critère.

## Division en trois groupes, pour étudier chaque rendez vous

J'aimerai m'intéresser plus particulièrement à chaque rendez vous, selon s'ils veulent se revoir ou pas.
Je place donc chaque personne et son partenaire dans un des trois groupes suivants:

*   Premier groupe Groupe_0_0 :les deux personnes du rendez vous ne veulent pas se revoir; il n'y a pas match.

*  Second groupe Groupe_0_1 : il n'y a qu'une personne sur les deux qui veut revoir l'autre; il n'y a pas match.

*  Troisième groupe Groupe_1_1: les deux personnes veulent se revoir; il y a match.


```python
dataset.shape
```

```python
dataset = dataset.dropna(subset=['pid'])
dataset['pid']=dataset['pid'].astype(int)


dataset_q1=pd.DataFrame()
dataset_q2=pd.DataFrame()
dataset_q3=pd.DataFrame()

for i in range (len(dataset)):
  iid_i=dataset.iloc[i]['iid']
  pid_i=dataset.iloc[i]['pid']
  ligne_i=dataset.iloc[[i]]
  row_pid = dataset[(dataset['iid'] == pid_i) & (dataset['pid']==iid_i)]
  if (row_pid.iloc[0]['dec_o']==0) and (ligne_i.iloc[0]['dec_o']==0):
    dataset_q1=pd.concat([dataset_q1,ligne_i,row_pid],ignore_index=True)
  elif (row_pid.iloc[0]['dec_o']==1) and (ligne_i.iloc[0]['dec_o']==1):
    dataset_q3=pd.concat([dataset_q3,ligne_i,row_pid],ignore_index=True)
  else:
    dataset_q2=pd.concat([dataset_q2,ligne_i,row_pid],ignore_index=True)

dataset_q1 = dataset_q1.drop_duplicates()
dataset_q2 = dataset_q2.drop_duplicates()
dataset_q3 = dataset_q3.drop_duplicates()
```

Je m'intéresse à la colonne int_corr, qui indique l'indice de corrélation entre les centres d'interêts

```python
df_corr_q1=dataset_q1['int_corr']
df_corr_q1 = df_corr_q1.iloc[::2].reset_index(drop=True) ## on supprime les corrélations doublées du partenaire
df_corr_q1 = df_corr_q1[df_corr_q1 >= 0]

df_corr_q2=dataset_q2['int_corr']
df_corr_q2 = df_corr_q2.iloc[::2].reset_index(drop=True)
df_corr_q2 = df_corr_q2[df_corr_q2 >= 0]

df_corr_q3=dataset_q3['int_corr']
df_corr_q3 = df_corr_q3.iloc[::2].reset_index(drop=True)
df_corr_q3 = df_corr_q3[df_corr_q3 >= 0]

df_box = pd.DataFrame({
    'corr': pd.concat([df_corr_q1, df_corr_q2, df_corr_q3]),
    'groupe': (
        ['Groupe_0_0'] * len(df_corr_q1) +
        ['Groupe_0_1'] * len(df_corr_q2) +
        ['Groupe_1_1'] * len(df_corr_q3)
    )
})

fig = px.box(
    df_box,
    x='groupe',
    y='corr',
    color='groupe',
    title="Distribution des corrélations d'intérêts par groupe"
)

fig.update_layout(
    xaxis_title="Groupes",
    yaxis_title="Corrélation d'intérêts"
)

fig.show()
```

Dans les trois groupes, on voit que la distribution de la corrélation des interêts est la même. On peut dire que ça n'a pas d'incidence.

## Activités et centres d'interêts communs au sein de chaque groupe

On s'intéresse maintenant tout de même aux réponses fournies dans le questionnaire sur les différents centres d'interêts et activités, et si cela a une influence selon les groupes.

```python
df_q1_pairs=dataset_q1.iloc[::2].reset_index(drop=True)      # On découpe en dataset pair et impair,
df_q1_impairs=dataset_q1.iloc[1::2].reset_index(drop=True)   # pour avoir chaque ligne des deux sous datasets qui
                                                             # correspondent à une personne du même rendez vous
df_q2_pairs=dataset_q2.iloc[::2].reset_index(drop=True)
df_q2_impairs=dataset_q2.iloc[1::2].reset_index(drop=True)

df_q3_pairs=dataset_q3.iloc[::2].reset_index(drop=True)
df_q3_impairs=dataset_q3.iloc[1::2].reset_index(drop=True)


## Je définis une fonction qui va compter pour chaque rendez vous une somme:
## +1 si la différence des notes attribuées de chaque partenaire pour le centre d'interêts est inférieure ou égale à 3
## 0 sinon, je stocke chaque somme de chaque rendez vous dans une liste

colonnes_interets = [
    'sports', 'tvsports', 'exercise', 'dining',
    'museums', 'art', 'hiking', 'gaming',
    'clubbing', 'reading', 'tv', 'theater',
    'movies', 'concerts', 'music', 'shopping', 'yoga'
]

def calcul_interets_communs(df_pairs, df_impairs):
    count_interest = []
    for i in range(len(df_pairs)):
        count = 0
        for col in colonnes_interets:
            if abs(df_pairs.loc[i, col] - df_impairs.loc[i, col]) <= 2:
                    count=count+1

        count_interest.append(count)

    # je ramène à un score sur 10 car il y a 17 centres d'interêts
    count_interest_sur10 = [(x / 17) * 10 for x in count_interest]

    return count_interest_sur10


q1_score = calcul_interets_communs(df_q1_pairs, df_q1_impairs)
q2_score = calcul_interets_communs(df_q2_pairs, df_q2_impairs)
q3_score = calcul_interets_communs(df_q3_pairs, df_q3_impairs)
```

```python
# Je stocke dans un dataframe pour comparer les statistiques

df_interest = pd.DataFrame({
    "groupe": (
        ["Groupe_0_0"] * len(q1_score) +
        ["Groupe_0_1"] * len(q2_score) +
        ["Groupe_1_1"] * len(q3_score)
    ),
    "valeur": q1_score + q2_score + q3_score
})

fig = px.box(
    df_interest,
    x="groupe",
    y="valeur",
    color="groupe",
    title="Distribution des centres d’intérêt en communs pour chaque groupe"
)

fig.update_traces(boxmean=True)

fig.update_layout(
    xaxis_title="Groupes",
    yaxis_title="Score sur 10"
)

fig.show()
```

On peut voir que la distribution est sensiblement la même dans les trois groupes, et que la médiane du groupe des matchs est un peu plus élevée, donc que les personnes ont un peu plus de centres d'interêts en communs. Ce n'est pas significatif non plus, on ne peut pas conclure que les centres d'interêts ont une influence dans les matchs.

## Observation des différences d'âge selon le groupe

```python
# On observe la différence d'âge entre chaque partenaire

diff_age_q1 = abs(df_q1_pairs.iloc[:, 7] - df_q1_impairs.iloc[:, 7])
diff_age_q2 = abs(df_q2_pairs.iloc[:, 7] - df_q2_impairs.iloc[:, 7])
diff_age_q3 = abs(df_q3_pairs.iloc[:, 7] - df_q3_impairs.iloc[:, 7])



df_age_mean = pd.DataFrame({
    "groupe": ["Groupe_0_0", "Groupe_0_1", "Groupe_1_1"],
    "moyenne_diff_age": [
        diff_age_q1.mean(),
        diff_age_q2.mean(),
        diff_age_q3.mean()
    ]
})

import plotly.express as px

fig = px.bar(
    df_age_mean,
    x="groupe",
    y="moyenne_diff_age",
    text="moyenne_diff_age",
    title="Moyenne des différences d'âge selon le groupe"
)

fig.update_layout(
    xaxis_title="Groupes",
    yaxis_title="Moyenne des différences d'âge"
)

fig.show()
```

Dans le groupe où il y a les matchs, les personnes ont en moyenne un peu plus le même âge que sur les deux autres groupes.

## Observation des études selon le groupe

```python
def calc_etudes(df_pairs, df_impairs):
    count = 0

    for i in range(len(df_pairs)):
        if df_pairs.iloc[i, 26] == df_impairs.iloc[i, 26]:
            count = count+1

    total = len(df_pairs)
    different = total - count

    return count, different

q1_same, q1_diff = calc_etudes(df_q1_pairs, df_q1_impairs)
q2_same, q2_diff = calc_etudes(df_q2_pairs, df_q2_impairs)
q3_same, q3_diff = calc_etudes(df_q3_pairs, df_q3_impairs)

df_all = pd.DataFrame({
    "groupe": ["Groupe_0_0", "Groupe_0_1", "Groupe_1_1"],
    "même études": [q1_same, q2_same, q3_same],
    "différentes": [q1_diff, q2_diff, q3_diff]
})


fig = make_subplots(
    rows=1, cols=3,
    specs=[[{"type": "domain"}, {"type": "domain"}, {"type": "domain"}]],
    subplot_titles=["Groupe_0_0", "Groupe_0_1", "Groupe_1_1"]
)

fig.add_trace(go.Pie(
    labels=["Même études", "Études différentes"],
    values=[q1_same, q1_diff],
    name="Groupe_0_0"
), 1, 1)

fig.add_trace(go.Pie(
    labels=["Même études", "Études différentes"],
    values=[q2_same, q2_diff],
    name="Groupe_0_1"
), 1, 2)

fig.add_trace(go.Pie(
    labels=["Même études", "Études différentes"],
    values=[q3_same, q3_diff],
    name="Groupe_1_1"
), 1, 3)

fig.update_layout(title_text="Comparaison des études entre partenaires")

fig.show()
```

On peut observer qu'il y a un peu plus de personnes qui font les mêmes études dans le groupe des matchs, mais ce n'est pas un critère qui permet de tirer une conclusion sur les matchs, et que la plupart des personnes font des études différentes.

## Conclusion

On observe, concernant les matchs, que les personnes attribuant des notes élevées sur l'ensemble des critères veulent revoir le partenaire de manière générale, mais l'attractivité est tout de même importante pour les deux sexes : les personnes qui ne donnent pas une bonne note sur l'attractivité ne veulent en général pas revoir la personne, même si elles ont attribué des bonnes notes sur les autres critères.
